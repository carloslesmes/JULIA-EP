# Diccionarios en Julia

## Creación de diccionarios

Un diccionario está constituido por parejas (llave(key), valor(value))

```julia
D1 = Dict("a" => 1, "b" => 2, 1 => "a")
```

```julia
D1["c"] = 3          
D1
```

```julia
D1["c"] = "Hello"  
D1  
```

```julia
D1 = delete!(D1, "c")  
D1
```

```julia
D2 = Dict("a" => 1, "b" => 2, "c" => 3, "d" => 4, "e" => 5)
```

```julia
D2["a"]
```

```julia
keys(D2)
```

```julia
values(D2)
```

```julia
for (key, value) in D2
    println("$key => $value")
end
```

```julia
dic1 = Dict("a" => 1, "b" => 2)
dic2 = Dict("b" => 3, "c" => 4)

unir_dics = merge(dic1, dic2)  
```

# Dataframes

Un `dataframe` es una estructura de datos con filas y columnas similar a una hoja de cálculo.  Organiza los datos en una tabla bidimensional. Las columnas tienen nombres y contienen datos del mismo tipo. Con los dataframes se pueden manejar datos faltantes.

```julia
using DataFrames
```

## Creación de un dataframe

```julia
df1= DataFrame("Edad" => [20,21,19], "nombre" => ["Carlos", "Juliana", "Pedro"])
```
