## Experimento aleatorio

Es un experimento en el cual:

1. No se conoce la respuesta de antemano.
2. Se conocen todas las posibles respuestas.
3. Se puede repetir bajo las mismas condiciones

## Ejemplos:

1. Lanzar una moneda, dos monedas, tres monedas,..... 
2. Lanzar un dado, dos dados,...

## Definiciones básicas

**Espacio muestral**: es el conjunto de todos los posibles resultados de un experimento aleatorio. Lo denotaremos $\Omega$.

**Evento**: es cualquier subconjunto del espacio muestral.

## Función de probabilidad

Sea 
$$
\begin{align} 
P: & \mathscr{P}(\Omega) \to [0,1] \\
& A \mapsto P(A)  
\end{align}
$$ 

una función de los eventos en $[0,1]$, que a cada evento $A$ le asigna un número real $P(A)$ entre 0 y 1, tal que:

1. $0 \leq P(A) \leq 1, \forall A \in \mathscr{P}(\Omega)$ 
2. $P(\phi)=0, P(\Omega)=1$
3. Si $A \cap B =\phi$ entonces $P(A \cup B)=P(A)+P(B)$,

entonces $P$ se llama función de probabilidad.

## Ejemplo

Lanzar una moneda, la función de probabilidad es:

* $\phi \to 0$
* $\{C\} \to 0.36$
* $\{S\} \to 0.64$
* $\Omega \to 1$

Lanzar la moneda 10000, supongamos que salen 3629 caras y 6371 sellos. **Cuando un enunciado no diga nada acerca de las probabilidades de un experimento se asume equiprobabilidad**.

## Propiedades de la función de probabilidad

Sean $A$ y $B$ eventos, se cumple:

1. $P(A^c)=1-P(A)$
2. $P(A-B)=P(A)-P(A \cap B)$
3. $P(A \cup B)= P(A)+P(B)-P(A \cap B)$

## Cálculo de probabilidades

Asumamos, en una primera instancia, que $\Omega$ es <font color=red>finito</font>, es decir $\Omega=\{\omega_1,\omega_2,\cdots,\omega_n\}$, con $P(\omega_1)+P(\omega_2)+\cdots+P(\omega_n)=1$, tenemos dos casos:

1. Si $P(\omega_1)=P(\omega_2)=\cdots=P(\omega_n)=\frac{1}{n}$, entonces  $$ P(A)=\frac{\#(A)}{\#(\Omega)}=\frac{\text{casos favorables}}{\text{casos posibles}} $$.

2. Si $P(\omega_i) \neq P(\omega_j)$ para $i \neq j$, entonces si $A=\{a_1,a_2,\cdots,a_m\}$
$$
P(A)=\sum_{i=1}^{m}P(a_i)
$$

# El problema clásico del cumpleaños

“¿Cuál es la probabilidad de que al menos dos personas cumplan años el mismo día en grupo de n personas?" 

Una simulación en Julia para estimar la probabilidad:

```julia
using Random

function simulacion(n_personas, n_ensayos=100000)
    coincidencias = 0

    for _ in 1:n_ensayos
    cumple = rand(1:365, n_personas)  # Genera días de cumpleaños
        if length(unique(cumple)) < n_personas  # ver duplicados
            coincidencias += 1
        end
    end

    return coincidencias / n_ensayos  #  probabilidad estimada
end

# simulation para 60 personas
simulacion(60)
@time simulacion(60) # tiempo  de julia
```

## Probabilidad condicional

Sean $A$ y $B$ eventos, definimos
$$ 
P(A|B)=\frac{P(A \cap B)}{P(B)}, \quad P(B)\ne 0
$$

Se lee: "probabilidad de A dado B", significa "calcular la probabilidad de A sabiendo que B ya sucedió".

### Ejemplo:

Se lanza un dado, **si** sale un número par, ¿Cuál es la probabilidad de que sea un número primo?

$\Omega=\{1,2,3,4,5,6\}$, sean los eventos A y B.

A: sale primo, $A=\{2,3,5\}$, B: sale par, $B=\{2,4,6\}$ por lo tanto $A \cap B=\{2\}$ y así:

$$
P(A|B)=\frac{\frac{1}{6}}{\frac{3}{6}}=\frac{1}{3}
$$

## Definición:

Decimos que dos eventos $A$ y $B$ son <font color=blue>**independientes**</font> si se cumple:

1. $P(A \cap B)=P(A)P(B)$, o
2. $P(A|B)=P(A)$

## Teorema de la multiplicación:

Para dos eventos $A$ y $B$ se tiene, a partir de la definición:

$P(A \cap B)=P(A)P(B|A)$, 

para tres eventos se tiene:

$P(A \cap B \cap C)=P(A)P(B|A)P(C|A \cap B)$
