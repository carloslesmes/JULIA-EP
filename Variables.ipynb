# Clasificación de variables

Las variables son el insumo de la estadística. Se clasifican en dos:

1. Cualitativas o categóricas
2. Cuantitativas
    1. Discretas
    2. Continuas

Las variables cualitativas toman como valores cualidades o atributos. Las Variables Discretas toman valores en los enteros y las continuas toman valoes en un intervalo de los números reales.

## Ejemplos de variables cualitativas

1. Color
2. Estado civil
3. Género
4. Profesión
5. Número de celular
6. Número de la cédula

Son atributos o cualidades

## Ejemplos de variables discretas

1. Número de hijos por familia
2. Número de clientes que llegan por hora
3. Número de accidentes por mes

Las variables discretas son para CONTAR.

## Ejemplos de variables continuas

1. Longitud
2. Tiempo
3. Ingreso
4. Peso
5. Temperatura
6. Velocidad

Las variables continuas son para MEDIR. Tienen unidades.

# Escalas de medida

Para las variables cualitativas:

1. Escala nominal: género
2. Escala ordinal : cabo, sargento, coronel, general

Para las variables cuantitativas:

1. Escala de intervalo: el cero no indica ausencia absoluta. °C
2. Escala de razón: el cero indica ausencia absoluta.


Ejemplo: Edad:______ está en escala de razón. 

Ejemplo:  

1. 20 a 30 
2. 30 a 40 
3. 40 a 50

en escala ordinal.
